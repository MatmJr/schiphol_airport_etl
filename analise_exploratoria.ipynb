Uma forma de conververter todos os valores que estão na coluna "aircraftType"

```python
#Estava subindo Warning:

# for i in range(len(df)):
#     df["aircraftType"][i] = ast.literal_eval(df["aircraftType"][i])

df['aircraftType'] = df['aircraftType'].apply(ast.literal_eval)
```

```python
df["aircraftType"][0]['iataMain']
```

```python
df["modeloAviao"] = None
```

```python
for i in range(len(df)):
    iata_info = df["aircraftType"][i]
    if 'iataMain' in iata_info:
        # O valor que você quer atribuir
        valor_a_atribuir = df["aircraftType"][i]['iataMain']
        # Atribuição usando .loc[linha, coluna]
        df.loc[i, "modeloAviao"] = valor_a_atribuir
```

```python
df["modeloAviao"].value_counts()
```

Criando uma tabela personalizada

```python
modelos_freq = df["modeloAviao"].value_counts().reset_index()
modelos_freq.columns = ["Modelo de Avião", "Frequência"]
```

```python
modelos_freq.head(10)
```

Plotando o nosso primeiro gráfico

```python
import matplotlib.pyplot as plt

plt.bar(modelos_freq['Modelo de Avião'].head(10), modelos_freq['Frequência'].head(10))
```

Um pouco de estatística

```python
modelos_freq['Frequência Relativa'] = modelos_freq.Frequência/modelos_freq.Frequência.sum()*100
```
